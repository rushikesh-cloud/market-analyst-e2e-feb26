# Technical Agent V2

Separate validation surface for the dynamic, tool-driven technical-analysis worker built with LangChain `create_agent`. V1 remains unchanged.

```python
from pathlib import Path

from market_analyst.config.settings import Settings
from market_analyst.services.agents.technical_v2 import run_technical_analysis_agent_v2
from market_analyst.types.technical_v2 import TechnicalAnalysisV2Request, TechnicalIndicatorConfig
```

```python
settings = Settings()
request = TechnicalAnalysisV2Request(
    ticker="MSFT",
    period="6mo",
    interval="1d",
    indicators=[
        TechnicalIndicatorConfig(name="bollinger_bands", parameters={"window": 20, "num_std": 2}),
        TechnicalIndicatorConfig(name="rsi", parameters={"period": 14, "overbought": 70, "oversold": 30}),
        TechnicalIndicatorConfig(name="macd", parameters={"fast_period": 12, "slow_period": 26, "signal_period": 9}),
    ],
    question="Summarize trend, momentum, and the main technical risks.",
)
output_dir = Path("notebooks/outputs/technical_charts_v2")
```

```python
result = run_technical_analysis_agent_v2(settings, request, output_dir=output_dir)
result
```

```python
from IPython.display import Image, display

display(Image(filename=str(result.chart_path)))
print(result.answer)
```

```python
assert result.rating is None or 1 <= result.rating <= 100
assert result.chart_path.exists()
assert result.artifact.indicator_names
```
